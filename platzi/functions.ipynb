### Mathematical Functions for Data Science and AI

Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Styles

```python
print(plt.style.available)
plt.style.use('classic')
```

Linear Functions
$$ y = mx + b $$

```python
n = 100
m = -1
b = 3

def f(x):
    return m*x+b

x = np.linspace(-10, 10, num=n)
y = f(x)

fig, ax = plt.subplots()
ax.plot(x,y,'b',label="$y = -x + 3$")
ax.set_title('Linear Function')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend(loc='lower left', bbox_to_anchor=(1,0.2))
ax.axhline(y=5, color='r')
ax.axvline(x=0, color='r')
ax.grid()
```

Polynomial Functions
$$y = x^7$$

```python
def f(x):
    return x**7

y = f(x)

plt.plot(x,y,'b',label="$y = x^7$")
plt.title('Polynomial Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left', bbox_to_anchor=(1,0.2))
plt.grid()
plt.axhline(y=0, color='r', linestyle='--')
plt.axvline(x=0, color='g', linestyle='--')
plt.show
```

Transcendental Functions
$$ y = cos(x) $$
$$ y = e^x $$
$$ y = log(x) $$

```python
def f(x):
    return np.cos(x)
y = f(x)
plt.plot(x,y,'b',label="$y = cos(x)$")
plt.title('Cosine Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left', bbox_to_anchor=(1,0.2))
plt.grid()
plt.show
```

```python
def f(x):
    return np.exp(x)
y = f(x)
plt.plot(x,y,'b',label="$y = e^x$")
plt.title('Exponential Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left', bbox_to_anchor=(1,0.2))
plt.grid()
plt.show
```

```python
def f(x):
    return np.log2(x)
y = f(x)
plt.plot(x,y,'b',label="$y = log(x)$")
plt.title('Logarithmic Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left', bbox_to_anchor=(1,0.2))
plt.grid()
plt.show
```

Step Functions

```python
def H(x):
    y = np.zeros(len(x))
    for idx, x in enumerate(x):
        if x >= 0:
            y[idx] = 1
    return y

n = 1000
x = np.linspace(-10,10,num=n)
y = H(x)

plt.plot(x,y,'b')
plt.title('Step Function')
plt.xlabel('x')
plt.ylabel('y')
plt.ylim(-0.2, 1.2)
plt.grid()
```

Composite Function

```python
def g(x):
    return x**2

def f(x):
    return np.sin(x)

y = g(x)

f_o_g = f(g(x))

plt.plot(x,f_o_g,'b',label="$y = sin(x^2)$")
plt.title('Composite Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left', bbox_to_anchor=(1,0.2))
plt.grid()
```
